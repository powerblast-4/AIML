```python
#boston housing

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn import linear_model
from sklearn.metrics import mean_squared_error,mean_absolute_error
import zipfile

# Corrected to explicitly open the zip file and read the specific CSV
with zipfile.ZipFile('/content/archive (4).zip', 'r') as z:
    with z.open('House_Rent_Dataset.csv') as f:
        df = pd.read_csv(f)

print(df)
print(df.info())
print(df.isnull().sum())
print(df.head())

plt.scatter(df['Size'],df['Rent'],color='red',marker='*') # Corrected column name from 'Built_up_area' to 'Size'
plt.xlabel('Size')
plt.ylabel('Rent')
plt.title('Scatter Plot')
plt.show()

# Explicitly selecting desired numeric feature columns for 'x' (BHK, Size, Bathroom)
x=df[['BHK', 'Size', 'Bathroom']]
y=df['Rent']
print(x)
print(y)

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42) # Corrected typo random_stae to random_state
print("xtrain shape:",x_train.shape)
print("xtest shape:",x_test.shape)
print("ytrain shape:",y_train.shape)
print("ytest shape:",y_test.shape)

reg=linear_model.LinearRegression()
reg.fit(x_train,y_train) # Train the model on the training data
y_pred=reg.predict(x_test)

# The predict line reg.predict([[650,0,0]]) was removed as its context was unclear with the dynamic 'x' features.
# If a specific prediction is desired, the input array should match the feature columns of 'x'.

print("Coefficients:", reg.coef_)
print("Intercept:", reg.intercept_)

# The hardcoded 'z' calculation is not relevant to the model's output and was removed.

train_r_squared=reg.score(x_train, y_train) # Calculate R-squared for training set
print("x_train (R-Squared):".replace('x_train', 'Training R-Squared'),train_r_squared)
test_r_squared=reg.score(x_test, y_test) # Calculate R-squared for test set
print("x_test (R-Squared):".replace('x_test', 'Test R-Squared'),test_r_squared)

mse=mean_squared_error(y_test,y_pred)
print("mean squared error:",mse)
mae=mean_absolute_error(y_test,y_pred)
print("mean absolute error:",mae)
```
